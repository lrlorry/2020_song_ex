```
# Detect GPU architecture
import subprocess
result = subprocess.run(['nvidia-smi', '--query-gpu=compute_cap', '--format=csv,noheader'],
                        capture_output=True, text=True)
cap = result.stdout.strip().replace('.', '')   # e.g. '7.5' -> '75'
arch = f'sm_{cap}'
print(f'GPU compute capability: {result.stdout.strip()}  →  -arch={arch}')
```

```
# Compile GPU binary
!nvcc -std=c++17 -O2 -arch={arch} -o song_gpu main.cu && echo 'GPU build OK'
```

```
# Compile CPU binary
!g++ -std=c++17 -O2 -o song_cpu main.cpp && echo 'CPU build OK'
```
